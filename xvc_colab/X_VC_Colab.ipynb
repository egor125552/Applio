1. Включи T4 GPU.
2. Нажми Выполнить всё.
3. Дождись публичной Gradio-ссылки.
4. Source Audio — исходная речь. Reference Audio — голос, который нужно перенести.
5. Для первого русского теста оставь Offline quality.

```python
from pathlib import Path
import subprocess, sys, urllib.request
TOOLS=Path('/content/xvc-colab-tools'); TOOLS.mkdir(parents=True, exist_ok=True)
BASE='https://raw.githubusercontent.com/egor125552/Applio/agent/xvc-gradio/xvc_colab'
for filename in ('install_xvc.py','requirements-colab.txt','app.py','launch_xvc.py'):
    urllib.request.urlretrieve(f'{BASE}/{filename}', TOOLS/filename)
subprocess.run([sys.executable, str(TOOLS/'install_xvc.py')], check=True)
```
